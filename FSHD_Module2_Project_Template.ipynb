* Diagnosis: There are two types of FSHD, FSHD1 and FSHD2. Both types manifest with the same symptoms, with the main difference being from the genetic cause. FSHD1 is typically associated with a shortened D4Z4 repeat array on chromosome 4, whereas FSHD2 involves a hypomethylation of the D4Z4 repeat array due to mutation on the SMCHD1, DNMT3B, or LRIF1 genes. FSHD1 is the most common form of the disease, accounting for 95% of cases, and FSHD is the rarer form accounting for 5% of cases.

* Standard of care treatments (& reimbursement): There is currently no established cure for FSHD. Management to decrease progression focuses on maintaining mobility and quality of life thorugh physical therapy, weight training, mobility aids, and pain managements. Some patients also undergo surgery to stabilize the scapula and regain range of motion. Since current approaches are tailured towards managing symptoms insteead of reversing muscle regeneration, there remains an unmet need for treatments that target the underlying disease mechanism. * Disease progression & prognosis: FSHD generally progresses gradually, with the rate and severaity varying substantially between patients and even between indiviudal muscles. Muscle weakness oftentimes begins in the face, shoulders, and upper arms before spreadng to other muscle groups. Over time, affected muscles undergo atrophy, fatty infiltration (replacement of muscle tissue with fat), fibrosis, and chronic inflammation. Although life expectancy is generally normal, some patients develop significant mobility limitations and may eventually require assistive devices. MRI-based muscle fat fraction can quantify fatty replacement, making it a potentially valuable biomarker for tracking disease progression and detecting changes that may not yet be apparent through physical tests.

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology): FSHD is caused by abnormal expression of the DUX4 gene in skeletal muscle. In mature skeletal muscle, DUX4 is normally suppressed, but genetic and epigenetic changes can cause it to become abnormally active. In FSHD1, this is primarily associated with contraction of the D4Z4 repeat region on chromosome 4, while FSHD2 results from changes that disrupt normal repression of this region. Abnormal DUX4 expression activates harmful molecular pathways that cause cellular stress, apoptosis (programmed cell death), and impaired muscle regeneration. Over long periods of time, these processes contribute to muscle fiber degeneration, atrophy, fatty infiltration, and fibrosis, reducing the muscle's ability to generate force. Because skeletal muscle contraction depends on interactions between actin and myosin within muscle fibers, the loss of functional muscle tissue contributes to progressive weakness.

Early Phase 1/2 clinical trials primarily assess treatment safety and  along with investigating endpoints that may indicate early biological or functional improvements. Potential endpoints include biomarkers such as DUX4-related gene-expression signatures, MRI-based muscle fat fraction, quantitative muscle strength measurements, or physical assessments such as walking and stair-climbing tests. However, since FSHD often progresses slowly and varies significantly between patients, identifying measurements that reliably detect disease progression over short periods remains a challenge in the field.

# 1. Import packages and locate the data
We imported Numpy, Pandas, Matplotlib, and a path so we could locate the data. We updated DATA_DIR to go through the correct folder containing our MRI data. This produced a table with the the columns subject, timepoint, slice_id, percent_along_muscle, z_index, distance_from_muscle_start_mm, water_file, fat_file, and mask file for each sampled slice. 
 

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("FSHD_MRI_data") / "FSHD_TA_MRI_data_0020_TP1_TP2"
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")
manifest
```

Table 1. MRI sample information for subject UW_0020

## 2. Load one sample

We used slice 4 from TP1 and used np.load() to load its water, fat, and mask files. We converted the mask to a Boolean value. This identifed which pixels were inside the TA muscle. This produced the dimensions of each image and the mask, which were all 266 × 224 pixels, and a muscle pixels of 360.


```python
row = manifest[manifest["timepoint"] == "TP1"].iloc[3]  # representative mid-muscle slice

water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
```

## 3. Calculate fat fraction

We calculated fat fraction for each pixel by dividing the fat signal by the combined water and fat signal and multiplying by 100. Then, we used the mask to separate the TA muscle from the surrounding region. This produced mean, minimum, and maximum fat fractions for both regions. The whole supplied image region mean was 48.6%, while the TA muscle only mean was 29.9%. The minimum for both was 0.0%. The maximum for the whole supplied image region was 100.0%, thile the maximum for the TA muscle only was 90.7%. 

```python
denominator = water + fat
valid_pixels = denominator > 0

fat_fraction = np.divide(
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100

# Keep the whole supplied image region for the first analysis.
fat_fraction_full = fat_fraction.copy()

# Then apply the TA mask for the muscle-specific analysis.
fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_masked[~np.isnan(fat_fraction_masked)]

print(f"Whole supplied image region — mean: {np.mean(whole_image_values):.1f}%")
print(f"Whole supplied image region — minimum: {np.min(whole_image_values):.1f}%")
print(f"Whole supplied image region — maximum: {np.max(whole_image_values):.1f}%")
print()
print(f"TA muscle only — mean: {np.mean(muscle_values):.1f}%")
print(f"TA muscle only — minimum: {np.min(muscle_values):.1f}%")
print(f"TA muscle only — maximum: {np.max(muscle_values):.1f}%")
```

## 4. Visualize the images, the mask, and fat-fraction map
We used np.where(mask) to find the muscle boundaries (min and max). The, we cropped its fat-fraction map. We usedplt.subplots() to create four images: water image with TA mask outline, fat image with TA mask outline, fat fraction for the whole region, and zoomed in fat fraction for the TA only. We labeled the x-axis as columns and the y-axis as rows, both in pixels. Both fat-fraction plots used the same 0–100% color scale. 

```python
# Find the bounding box of the TA mask
y_pixels, x_pixels = np.where(mask)

y0, y1 = y_pixels.min(), y_pixels.max() + 1
x0, x1 = x_pixels.min(), x_pixels.max() + 1

# Crop the TA-specific fat-fraction map
fat_fraction_TA = fat_fraction_masked[y0:y1, x0:x1]

fig, axes = plt.subplots(1, 4, figsize=(15, 4))

# Water image with TA mask outline
axes[0].imshow(water, cmap="gray")
axes[0].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[0].set_title("Water image + TA mask")

# Fat image with TA mask outline
axes[1].imshow(fat, cmap="gray")
axes[1].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[1].set_title("Fat image + TA mask")

# Fat fraction for the whole supplied image region
whole_image = axes[2].imshow(
    fat_fraction_full,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[2].contour(mask, levels=[0.5], colors="cyan", linewidths=2)
axes[2].set_title("Fat fraction: whole region")

# Zoomed-in fat fraction for the TA only
ta_image = axes[3].imshow(
    fat_fraction_TA,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[3].set_title("TA fat fraction")

for ax in axes:
    ax.set_xlabel("Column (pixels)")
    ax.set_ylabel("Row (pixels)")

# Both fat-fraction plots use the same 0–100% scale
fig.colorbar(
    ta_image,
    ax=axes[2:],
    fraction=0.046,
    pad=0.04,
    label="Fat fraction (%)"
)

plt.show()
```

Figure 1. MRI images and fat-fraction plots for TP1 slice 4. 

The red and blue outlines show where the TA muscle is. The zoomed in image (4th image) shows how the fat fraction canges within the muscle. The higher values are yellow and toward the edges. 


```python
plt.figure(figsize=(7, 4))
plt.hist(whole_image_values, bins=30, alpha=0.55, label="Whole supplied region")
plt.hist(muscle_values, bins=30, alpha=0.65, label="TA muscle only")
plt.xlabel("Fat fraction (%)")
plt.ylabel("Number of pixels")
plt.title("Fat-fraction distributions")
plt.legend()
plt.show()
```

```python
def calculate_fat_fraction(water, fat, mask):
    """Return a masked fat-fraction image in percent."""
    denominator = water + fat
    result = np.divide(
        fat,
        denominator,
        out=np.zeros_like(fat, dtype=float),
        where=denominator > 0
    ) * 100
    result[~mask] = np.nan
    return result

analysis_timepoint = "TP1"  # analyze one visit at a time
analysis_manifest = manifest[manifest["timepoint"] == analysis_timepoint]
results = []

for _, row in analysis_manifest.iterrows():
    water = np.load(DATA_DIR / row["water_file"])
    fat = np.load(DATA_DIR / row["fat_file"])
    mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
    ff = calculate_fat_fraction(water, fat, mask)

    results.append({
        "subject": row["subject"],
        "timepoint": row["timepoint"],
        "slice_id": row["slice_id"],
        "percent_along_muscle": row["percent_along_muscle"],
        "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
        "mean_fat_fraction_percent": np.nanmean(ff),
        "median_fat_fraction_percent": np.nanmedian(ff),
        "std_fat_fraction_percent": np.nanstd(ff),
        "n_muscle_pixels": int(mask.sum()),
    })

results_df = pd.DataFrame(results)
results_df
```

```python
def summarize_visit(timepoint):
    visit_results = []
    visit_manifest = manifest[manifest["timepoint"] == timepoint]

    for _, row in visit_manifest.iterrows():
        water = np.load(DATA_DIR / row["water_file"])
        fat = np.load(DATA_DIR / row["fat_file"])
        mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
        ff = calculate_fat_fraction(water, fat, mask)

        visit_results.append({
            "timepoint": timepoint,
            "slice_id": row["slice_id"],
            "percent_along_muscle": row["percent_along_muscle"],
            "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
            "mean_fat_fraction_percent": np.nanmean(ff),
        })

    return pd.DataFrame(visit_results)

profiles_df = pd.concat(
    [summarize_visit("TP1"), summarize_visit("TP2")],
    ignore_index=True
)

plt.figure(figsize=(7, 4))
for timepoint, group in profiles_df.groupby("timepoint"):
    group = group.sort_values("distance_from_muscle_start_mm")
    plt.plot(
        group["percent_along_muscle"],
        group["mean_fat_fraction_percent"],
        marker="o",
        label=timepoint
    )
plt.xlabel("Distance from muscle start (mm)")
plt.ylabel("Mean TA fat fraction (%)")
plt.title("Mean fat fraction along the tibialis anterior")
plt.legend(title="Visit")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
output_file = Path("TA_fat_fraction_summary.csv")
results_df.to_csv(output_file, index=False)
print(f"Saved {output_file}")
```
