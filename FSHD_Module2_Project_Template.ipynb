## Project Goal:

Facioscapulohumeral muscular dystrophy (FSHD) is a genetic disorder characterized by progressive skeletal muscle weakness and degeneration. A major challenge in developing treatments is identifying reliable measurements that detect disease progression over relatively short periods, particularly in early clinical trials evaluating potential therapies.

This project aims to analyze natural-history data from patients with FSHD to compare MRI-based muscle fat fraction, quantitative muscle strength, functional assessments, and clinical severity scores as potential exploratory endpoints for an early Phase 1/2 clinical trial.

* Prevalence & incidence: FSHD has an estimated prevalence of approximately 4–10 individuals per 100,000 people, although estimates vary between populations. A population-based study in the Netherlands from 2001 to 2010 estimated an annual incidence of 0.3 newly diagnosed individuals per 100,000 person, which corresponded to approximately 52 new diagnoses per year in that population. Since FSHD symptoms vary in severity and some affected individuals remain undiagnosed, it's possible that its true prevalence may be underestimated. 

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/
https://pmc.ncbi.nlm.nih.gov/articles/PMC4166358/


* Symptoms: FSHD causes progressive skeletal muscle weakness, particularly affecting the facial muscles, shoulders, upper arms, and lower legs. Common symptoms include difficulty raising the arms, walking, lifting the feet, alongside fatigue and chronic pain. Muscle weakness is frquently asymmetric, meaning one side of the body may be more severely affected than the other.

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/
https://medlineplus.gov/genetics/condition/facioscapulohumeral-muscular-dystrophy/
https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms



* Diagnosis: There are two types of FSHD, FSHD1 and FSHD2. Both types manifest with the same symptoms, with the main difference being from the genetic cause. FSHD1 is typically associated with a shortened D4Z4 repeat array on chromosome 4, whereas FSHD2 involves a hypomethylation of the D4Z4 repeat array due to mutation on the SMCHD1, DNMT3B, or LRIF1 genes. FSHD1 is the most common form of the disease, accounting for 95% of cases, and FSHD is the rarer form accounting for 5% of cases. 

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/


* Standard of care treatments (& reimbursement): There is currently no established cure for FSHD. Management to decrease progression focuses on maintaining mobility and quality of life thorugh physical therapy, weight training, mobility aids, and pain managements. Some patients also undergo surgery to stabilize the scapula and regain range of motion. Since current approaches are tailured towards managing symptoms insteead of reversing muscle regeneration, there remains an unmet need for treatments that target the underlying disease mechanism. Since there is no expensive drug treament, cost of management is mostly dpeendent on insurance coverage. 

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/
https://pmc.ncbi.nlm.nih.gov/articles/PMC4520817/


* Disease progression & prognosis: FSHD generally progresses gradually, with the rate and severaity varying substantially between patients and even between indiviudal muscles. Muscle weakness oftentimes begins in the face, shoulders, and upper arms before spreadng to other muscle groups. Over time, affected muscles undergo atrophy, fatty infiltration (replacement of muscle tissue with fat), fibrosis, and chronic inflammation. Although life expectancy is generally normal, some patients develop significant mobility limitations and may eventually require assistive devices. MRI-based muscle fat fraction can quantify fatty replacement, making it a potentially valuable biomarker for tracking disease progression and detecting changes that may not yet be apparent through physical tests. 

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/
https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms


* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology): FSHD is caused by abnormal expression of the DUX4 gene in skeletal muscle. In mature skeletal muscle, DUX4 is normally suppressed, but genetic and epigenetic changes can cause it to become abnormally active. In FSHD1, this is primarily associated with contraction of the D4Z4 repeat region on chromosome 4, while FSHD2 results from changes that disrupt normal repression of this region. Abnormal DUX4 expression activates harmful molecular pathways that cause cellular stress, apoptosis (programmed cell death), and impaired muscle regeneration. Over long periods of time, these processes contribute to muscle fiber degeneration, atrophy, fatty infiltration, and fibrosis, reducing the muscle's ability to generate force. Because skeletal muscle contraction depends on interactions between actin and myosin within muscle fibers, the loss of functional muscle tissue contributes to progressive weakness.

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/
https://www.fshdsociety.org/living-with-fshd/understanding-fshd/genetic-cause/


* Clinical Trials/next-gen therapies:  Emerging therapeutic approaches include reducing abnormal DUX4 expression or blocking its harmful downstream effects to prevent further muscle damage. Early Phase 1/2 clinical trials primarily assess treatment safety and  along with investigating endpoints that may indicate early biological or functional improvements. Potential endpoints include biomarkers such as DUX4-related gene-expression signatures, MRI-based muscle fat fraction, quantitative muscle strength measurements, or physical assessments such as walking and stair-climbing tests. However, since FSHD often progresses slowly and varies significantly between patients, identifying measurements that reliably detect disease progression over short periods remains a challenge in the field. 

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/
https://clinicaltrials.gov/search?cond=Facioscapulohumeral%20Muscular%20Dystrophy&viewType=Card
