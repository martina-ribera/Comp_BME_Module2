## Disease Background: 

* Prevalence & incidence: FSHD has an estimated prevalence of approximately 4–10 individuals per 100,000 people, although estimates vary between populations. A population-based study in the Netherlands from 2001 to 2010 estimated an annual incidence of 0.3 newly diagnosed individuals per 100,000 person, which corresponded to approximately 52 new diagnoses per year in that population. Since FSHD symptoms vary in severity and some affected individuals remain undiagnosed, it's possible that its true prevalence may be underestimated. 

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/
https://pmc.ncbi.nlm.nih.gov/articles/PMC4166358/


* Symptoms: FSHD causes progressive skeletal muscle weakness, particularly affecting the facial muscles, shoulders, upper arms, and lower legs. Common symptoms include difficulty raising the arms, walking, lifting the feet, alongside fatigue and chronic pain. Muscle weakness is frquently asymmetric, meaning one side of the body may be more severely affected than the other.

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/
https://medlineplus.gov/genetics/condition/facioscapulohumeral-muscular-dystrophy/
https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms



* Diagnosis: There are two types of FSHD, FSHD1 and FSHD2. Both types manifest with the same symptoms, with the main difference being from the genetic cause. FSHD1 is typically associated with a shortened D4Z4 repeat array on chromosome 4, whereas FSHD2 involves a hypomethylation of the D4Z4 repeat array due to mutation on the SMCHD1, DNMT3B, or LRIF1 genes. FSHD1 is the most common form of the disease, accounting for 95% of cases, and FSHD is the rarer form accounting for 5% of cases. 

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/


* Standard of care treatments (& reimbursement): There is currently no established cure for FSHD. Management to decrease progression focuses on maintaining mobility and quality of life thorugh physical therapy, weight training, mobility aids, and pain managements. Some patients also undergo surgery to stabilize the scapula and regain range of motion. Since current approaches are tailured towards managing symptoms insteead of reversing muscle regeneration, there remains an unmet need for treatments that target the underlying disease mechanism. Since there is no expensive drug treament, cost of management is mostly dpeendent on insurance coverage. 

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/
https://pmc.ncbi.nlm.nih.gov/articles/PMC4520817/


* Disease progression & prognosis: FSHD generally progresses gradually, with the rate and severaity varying substantially between patients and even between indiviudal muscles. Muscle weakness oftentimes begins in the face, shoulders, and upper arms before spreadng to other muscle groups. Over time, affected muscles undergo atrophy, fatty infiltration (replacement of muscle tissue with fat), fibrosis, and chronic inflammation. Although life expectancy is generally normal, some patients develop significant mobility limitations and may eventually require assistive devices. MRI-based muscle fat fraction can quantify fatty replacement, making it a potentially valuable biomarker for tracking disease progression and detecting changes that may not yet be apparent through physical tests. 

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/
https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms


* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology): FSHD is caused by abnormal expression of the DUX4 gene in skeletal muscle. In mature skeletal muscle, DUX4 is normally suppressed, but genetic and epigenetic changes can cause it to become abnormally active. In FSHD1, this is primarily associated with contraction of the D4Z4 repeat region on chromosome 4, while FSHD2 results from changes that disrupt normal repression of this region. Abnormal DUX4 expression activates harmful molecular pathways that cause cellular stress, apoptosis (programmed cell death), and impaired muscle regeneration. Over long periods of time, these processes contribute to muscle fiber degeneration, atrophy, fatty infiltration, and fibrosis, reducing the muscle's ability to generate force. Because skeletal muscle contraction depends on interactions between actin and myosin within muscle fibers, the loss of functional muscle tissue contributes to progressive weakness.

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/
https://www.fshdsociety.org/living-with-fshd/understanding-fshd/genetic-cause/


* Clinical Trials/next-gen therapies:  Emerging therapeutic approaches include reducing abnormal DUX4 expression or blocking its harmful downstream effects to prevent further muscle damage. Early Phase 1/2 clinical trials primarily assess treatment safety and  along with investigating endpoints that may indicate early biological or functional improvements. Potential endpoints include biomarkers such as DUX4-related gene-expression signatures, MRI-based muscle fat fraction, quantitative muscle strength measurements, or physical assessments such as walking and stair-climbing tests. However, since FSHD often progresses slowly and varies significantly between patients, identifying measurements that reliably detect disease progression over short periods remains a challenge in the field. 

Source: https://www.ncbi.nlm.nih.gov/books/NBK1443/

# 1. Import packages and locate the data
We imported Numpy, Pandas, Matplotlib, and a path so we could locate the data. We updated DATA_DIR to go through the correct folder containing our MRI data. This produced a table with the the columns subject, timepoint, slice_id, percent_along_muscle, z_index, distance_from_muscle_start_mm, water_file, fat_file, and mask file for each sampled slice. 
 

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("FSHD_MRI_data") / "FSHD_TA_MRI_data_0020_TP1_TP2"
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")
manifest
```

Table 1. MRI sample information for subject UW_0020

## 2. Load one sample

We used slice 4 from TP1 and used np.load() to load its water, fat, and mask files. We converted the mask to a Boolean value. This identifed which pixels were inside the TA muscle. This produced the dimensions of each image and the mask, which were all 266 × 224 pixels, and a muscle pixels of 360.


```python
row = manifest[manifest["timepoint"] == "TP1"].iloc[3]  # representative mid-muscle slice

water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
```

## 3. Calculate fat fraction

We calculated fat fraction for each pixel by dividing the fat signal by the combined water and fat signal and multiplying by 100. Then, we used the mask to separate the TA muscle from the surrounding region. This produced mean, minimum, and maximum fat fractions for both regions. The whole supplied image region mean was 48.6%, while the TA muscle only mean was 29.9%. The minimum for both was 0.0%. The maximum for the whole supplied image region was 100.0%, thile the maximum for the TA muscle only was 90.7%. 

```python
denominator = water + fat
valid_pixels = denominator > 0

fat_fraction = np.divide(
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100

# Keep the whole supplied image region for the first analysis.
fat_fraction_full = fat_fraction.copy()

# Then apply the TA mask for the muscle-specific analysis.
fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_masked[~np.isnan(fat_fraction_masked)]

print(f"Whole supplied image region — mean: {np.mean(whole_image_values):.1f}%")
print(f"Whole supplied image region — minimum: {np.min(whole_image_values):.1f}%")
print(f"Whole supplied image region — maximum: {np.max(whole_image_values):.1f}%")
print()
print(f"TA muscle only — mean: {np.mean(muscle_values):.1f}%")
print(f"TA muscle only — minimum: {np.min(muscle_values):.1f}%")
print(f"TA muscle only — maximum: {np.max(muscle_values):.1f}%")
```

## 4. Visualize the images, the mask, and fat-fraction map
We used np.where(mask) to find the muscle boundaries (min and max). The, we cropped its fat-fraction map. We usedplt.subplots() to create four images: water image with TA mask outline, fat image with TA mask outline, fat fraction for the whole region, and zoomed in fat fraction for the TA only. We labeled the x-axis as columns and the y-axis as rows, both in pixels. Both fat-fraction plots used the same 0–100% color scale. 

```python
# Find the bounding box of the TA mask
y_pixels, x_pixels = np.where(mask)

y0, y1 = y_pixels.min(), y_pixels.max() + 1
x0, x1 = x_pixels.min(), x_pixels.max() + 1

# Crop the TA-specific fat-fraction map
fat_fraction_TA = fat_fraction_masked[y0:y1, x0:x1]

fig, axes = plt.subplots(1, 4, figsize=(15, 4))

# Water image with TA mask outline
axes[0].imshow(water, cmap="gray")
axes[0].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[0].set_title("Water image + TA mask")

# Fat image with TA mask outline
axes[1].imshow(fat, cmap="gray")
axes[1].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[1].set_title("Fat image + TA mask")

# Fat fraction for the whole supplied image region
whole_image = axes[2].imshow(
    fat_fraction_full,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[2].contour(mask, levels=[0.5], colors="cyan", linewidths=2)
axes[2].set_title("Fat fraction: whole region")

# Zoomed-in fat fraction for the TA only
ta_image = axes[3].imshow(
    fat_fraction_TA,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[3].set_title("TA fat fraction")

for ax in axes:
    ax.set_xlabel("Column (pixels)")
    ax.set_ylabel("Row (pixels)")

# Both fat-fraction plots use the same 0–100% scale
fig.colorbar(
    ta_image,
    ax=axes[2:],
    fraction=0.046,
    pad=0.04,
    label="Fat fraction (%)"
)

plt.show()
```

## 6. Repeat the analysis for every sampled location in one visit

We created a function called calculate_fat_fraction() to repeat the calculation for each slice. We used a forloop repeast it for all eight TP1 slices. Then we calculated the mean, median, and standard deviation of TA fat fraction and the muscle pizels per slice. We used pd.DataFrame() to create a table.

```python
def calculate_fat_fraction(water, fat, mask):
    """Return a masked fat-fraction image in percent."""
    denominator = water + fat
    result = np.divide(
        fat,
        denominator,
        out=np.zeros_like(fat, dtype=float),
        where=denominator > 0
    ) * 100
    result[~mask] = np.nan
    return result

analysis_timepoint = "TP1"  # analyze one visit at a time
analysis_manifest = manifest[manifest["timepoint"] == analysis_timepoint]
results = []

for _, row in analysis_manifest.iterrows():
    water = np.load(DATA_DIR / row["water_file"])
    fat = np.load(DATA_DIR / row["fat_file"])
    mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
    ff = calculate_fat_fraction(water, fat, mask)

    results.append({
        "subject": row["subject"],
        "timepoint": row["timepoint"],
        "slice_id": row["slice_id"],
        "percent_along_muscle": row["percent_along_muscle"],
        "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
        "mean_fat_fraction_percent": np.nanmean(ff),
        "median_fat_fraction_percent": np.nanmedian(ff),
        "std_fat_fraction_percent": np.nanstd(ff),
        "n_muscle_pixels": int(mask.sum()),
    })

results_df = pd.DataFrame(results)
results_df
```

Table 2. TA fat-fraction measurements across eight TP1 slices

According to the table, the mean fat fraction decreased from 83.9% in slice 1 to 16.8% in slice 6, then increased to 27.9% in slice 8. This shows that the muscle has a higher fat fraction in some slices than in others.

## 7. Extension: compare fat fraction profiles across two visits

We created the function summarize_visit() to calculate mean TA fat fraction at each sampled location for TP1 and TP2. We used pd.concat() to combine the results and plt.plot() to graph. We plotted distance from the muscle start in millimeters on the x-axis and mean TA fat fraction (%) on the y-axis.

```python
def summarize_visit(timepoint):
    visit_results = []
    visit_manifest = manifest[manifest["timepoint"] == timepoint]

    for _, row in visit_manifest.iterrows():
        water = np.load(DATA_DIR / row["water_file"])
        fat = np.load(DATA_DIR / row["fat_file"])
        mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
        ff = calculate_fat_fraction(water, fat, mask)

        visit_results.append({
            "timepoint": timepoint,
            "slice_id": row["slice_id"],
            "percent_along_muscle": row["percent_along_muscle"],
            "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
            "mean_fat_fraction_percent": np.nanmean(ff),
        })

    return pd.DataFrame(visit_results)

profiles_df = pd.concat(
    [summarize_visit("TP1"), summarize_visit("TP2")],
    ignore_index=True
)

plt.figure(figsize=(7, 4))
for timepoint, group in profiles_df.groupby("timepoint"):
    group = group.sort_values("distance_from_muscle_start_mm")
    plt.plot(
        group["distance_from_muscle_start_mm"],
        group["mean_fat_fraction_percent"],
        marker="o",
        label=timepoint
    )
plt.xlabel("Distance from muscle start (mm)")
plt.ylabel("Mean TA fat fraction (%)")
plt.title("Mean fat fraction along the tibialis anterior")
plt.legend(title="Visit")
plt.grid(True, alpha=0.3)
plt.show()
```

Figure 3. Mean TA fat fraction along the muscle at TP1 and TP2. 

The blue represents TP1 and orange represents TP2 as seen in the legend. Both visits decrease and then increase slightly at the last sampled location. The orange line has higher mean TA fat fraction values at most locations, suggesting more fat in those parts of the muscle at the second visit.

## 5. Linear interpolation at T1

We used the baseline and Year 1 MRI fat fraction measurements for subject 13-0004 to estimate fat fraction at T1 (11.15 months). We constructed matrices Z and Y and used `np.linalg.solve()` to calculate the coefficients of a linear equation. This allowed us to estimate fat fraction between two measured visits.

```python
# Linear interpolation using baseline and Year 1
x1, x2 = lag_months[0], lag_months[1]
y1, y2 = ff_values[0], ff_values[1]

# Construct matrices
Z = np.array([[1, x1],
              [1, x2]])
Y = np.array([y1, y2])

# Solve for coefficients
A = np.linalg.solve(Z, Y)
a1, a2 = A

# Estimate fat fraction at T1
estimate_T1 = a1 + a2 * T1

print(f"Linear estimate at T1 = {T1:.2f} months: {estimate_T1:.2f}%")
```

## 6. Quadratic interpolation at T2

We used the baseline, Year 1, and Year 2 MRI fat fraction measurements for subject 13-0004 to estimate fat fraction at T2 (22.13 months). We constructed a matrix containing the three visit times and their squared values, then used `np.linalg.solve()` to determine the coefficients of a quadratic equation.

```python
# Quadratic interpolation using all three visits
Z = np.array([[1, x, x**2] for x in lag_months])
Y = ff_values

# Solve for coefficients
A = np.linalg.solve(Z, Y)
a1, a2, a3 = A

# Estimate fat fraction at T2
estimate_T2 = a1 + a2 * T2 + a3 * T2**2

print(f"Quadratic estimate at T2 = {T2:.2f} months: {estimate_T2:.2f}%")
```

## 7. Verify interpolation 

We used SciPy's `interp1d()` function to verify the linear and quadratic interpolation estimates calculated using the matrix method. Comparing the results allowed us to confirm that both approaches produced the same estimated fat fractions.

```python
# Create linear and quadratic interpolation functions
f_linear = interp1d(lag_months, ff_values, kind='linear')
f_quadratic = interp1d(lag_months, ff_values, kind='quadratic')

# Verify the estimates
print("Linear estimate at T1:", f_linear(T1))
print("Quadratic estimate at T2:", f_quadratic(T2))
```

We estimated right tibialis anterior fat fraction at T1 using linear interpolation and at T2 using quadratic interpolation. We accounted for missing measurements and avoided extrapolation by checking whether each subject's recorded visits covered the selected time points.

```python
results = []

for idx, row in df.iterrows():
    subject_id = row['subject_id']

    lag_months = np.array([row['lag_months_baseline'],
                           row['lag_months_y1'],
                           row['lag_months_y2']])

    ff_values = np.array([row['ff_baseline_R'],
                          row['ff_y1_R'],
                          row['ff_y2_R']])

    # Linear interpolation at T1
    estimate_T1 = np.nan

    if pd.notna(lag_months[0]) and pd.notna(lag_months[1]):
        if pd.notna(ff_values[0]) and pd.notna(ff_values[1]):
            if lag_months[0] < lag_months[1]:
                if lag_months[0] <= T1 <= lag_months[1]:

                    f_linear = interp1d(lag_months[:2],
                                        ff_values[:2], kind='linear')
                    estimate_T1 = f_linear(T1)

    # Quadratic interpolation at T2
    estimate_T2 = np.nan

    if pd.notna(lag_months[0]) and pd.notna(lag_months[1]) and pd.notna(lag_months[2]):
        if pd.notna(ff_values[0]) and pd.notna(ff_values[1]) and pd.notna(ff_values[2]):
            if lag_months[0] < lag_months[1] < lag_months[2]:
                if lag_months[0] <= T2 <= lag_months[2]:

                    f_quadratic = interp1d(lag_months,
                                           ff_values, kind='quadratic')
                    estimate_T2 = f_quadratic(T2)

    results.append({
        'subject_id': subject_id,
        'estimate_T1_linear': estimate_T1,
        'estimate_T2_quadratic': estimate_T2
    })

results_df = pd.DataFrame(results)
print(results_df)
```

## 12. Interpolate four-stair ascent measurements

We applied linear and quadratic interpolation to the four-stair ascent measurements for each subject. We estimated functional performance at the same common time points used for MRI fat fraction, allowing us to compare changes in both measurements over time. We excluded estimates requiring extrapolation or missing measurements.

```python
functional_results = []

for idx, row in df.iterrows():
    subject_id = row['subject_id']

    lag_months = np.array([row['lag_months_baseline'],
                           row['lag_months_y1'],
                           row['lag_months_y2']])

    ascent_values = np.array([row['ascend_4_baseline'],
                              row['ascend_4_y1'],
                              row['ascend_4_y2']])

    # Linear interpolation at T1
    ascent_T1 = np.nan

    if pd.notna(lag_months[0]) and pd.notna(lag_months[1]):
        if pd.notna(ascent_values[0]) and pd.notna(ascent_values[1]):
            if lag_months[0] < lag_months[1]:
                if lag_months[0] <= T1 <= lag_months[1]:

                    f_linear = interp1d(lag_months[:2],
                                        ascent_values[:2], kind='linear')
                    ascent_T1 = f_linear(T1)

    # Quadratic interpolation at T2
    ascent_T2 = np.nan

    if pd.notna(lag_months[0]) and pd.notna(lag_months[1]) and pd.notna(lag_months[2]):
        if pd.notna(ascent_values[0]) and pd.notna(ascent_values[1]) and pd.notna(ascent_values[2]):
            if lag_months[0] < lag_months[1] < lag_months[2]:
                if lag_months[0] <= T2 <= lag_months[2]:

                    f_quadratic = interp1d(lag_months,
                                           ascent_values, kind='quadratic')
                    ascent_T2 = f_quadratic(T2)

    functional_results.append({
        'subject_id': subject_id,
        'ascent_T1_linear': ascent_T1,
        'ascent_T2_quadratic': ascent_T2
    })

functional_df = pd.DataFrame(functional_results)
print(functional_df)
```

## 13. Check the available interpolation results

We counted the number of subjects with valid interpolated four-stair ascent measurements at T1 and T2. We then identified how many subjects had complete estimates for both MRI fat fraction and four-stair ascent, allowing us to use the same subjects when comparing disease progression.

```python
# Count available four-stair ascent estimates
print("Subjects with ascent T1 estimates:",
      functional_df['ascent_T1_linear'].notna().sum())

print("Subjects with ascent T2 estimates:",
      functional_df['ascent_T2_quadratic'].notna().sum())

# Count subjects with complete estimates for both measurements
complete_subjects = 0

for idx, row in results_df.iterrows():

    if pd.notna(row['estimate_T1_linear']):
        if pd.notna(row['estimate_T2_quadratic']):
            if pd.notna(functional_df['ascent_T1_linear'][idx]):
                if pd.notna(functional_df['ascent_T2_quadratic'][idx]):

                    complete_subjects = complete_subjects + 1

print("Subjects with all four estimates:", complete_subjects)
```

## 14. Calculate percentage change in MRI fat fraction and four-stair ascent

We calculated the percentage change in MRI fat fraction and four-stair ascent time between T1 and T2 for subjects with complete interpolated measurements. This allowed us to compare the relative progression detected by each measurement, despite their different units. Positive changes indicate increased muscle fat infiltration or longer stair ascent times, respectively.

```python
# Calculate percentage changes for subjects with complete measurements
mri_changes = []
ascent_changes = []

for idx, row in results_df.iterrows():

    if pd.notna(row['estimate_T1_linear']):
        if pd.notna(row['estimate_T2_quadratic']):
            if pd.notna(functional_df['ascent_T1_linear'][idx]):
                if pd.notna(functional_df['ascent_T2_quadratic'][idx]):

                    # MRI fat fraction percentage change
                    mri_T1 = row['estimate_T1_linear']
                    mri_T2 = row['estimate_T2_quadratic']

                    # Four-stair ascent percentage change
                    ascent_T1 = functional_df['ascent_T1_linear'][idx]
                    ascent_T2 = functional_df['ascent_T2_quadratic'][idx]

                    if mri_T1 != 0:
                        if ascent_T1 != 0:

                            mri_change = (mri_T2 - mri_T1) / mri_T1 * 100
                            ascent_change = (ascent_T2 - ascent_T1) / ascent_T1 * 100

                            mri_changes.append(mri_change)
                            ascent_changes.append(ascent_change)

print("MRI percentage changes:")
for change in mri_changes:
    print(f"{change:.2f}%")

print("Four-stair ascent percentage changes:")
for change in ascent_changes:
    print(f"{change:.2f}%")

print("Subjects included:", len(mri_changes))
```

## 15. Calculate mean and standard deviation of percentage changes

We calculated the mean percentage change and standard deviation for MRI fat fraction and four-stair ascent time across the 24 subjects with complete interpolated measurements. The mean represents the average change between T1 and T2, while the standard deviation represents variability between subjects. These results allow us to evaluate which measurement detects FSHD progression more consistently.

```python
# Calculate mean percentage changes
mri_mean = np.mean(mri_changes)
ascent_mean = np.mean(ascent_changes)

# Calculate standard deviations
mri_std = np.std(mri_changes, ddof=1)
ascent_std = np.std(ascent_changes, ddof=1)

print(f"MRI mean change: {mri_mean:.2f}%")
print(f"MRI standard deviation: {mri_std:.2f}%")

print(f"Four-stair ascent mean change: {ascent_mean:.2f}%")
print(f"Four-stair ascent standard deviation: {ascent_std:.2f}%")
```

## 16. Compare MRI fat fraction and four-stair ascent

We created a bar graph comparing the mean percentage changes in MRI fat fraction and four-stair ascent time between T1 and T2. We included standard deviation error bars to illustrate variability across the 24 subjects. This allows us to visually compare the progression signals and consistency of both potential clinical trial endpoints.

```python
# Mean percentage changes and standard deviations
means = [mri_mean, ascent_mean]
stds = [mri_std, ascent_std]

# Plot mean percentage change with standard deviation
plt.figure(figsize=(7, 5))

plt.bar(["MRI Fat Fraction", "Four-Stair Ascent"],
        means, yerr=stds, capsize=5)

plt.ylabel("Mean Percentage Change (%)")
plt.title("FSHD Progression: MRI vs. Functional Assessment")
plt.grid(True, axis="y", alpha=0.3)
plt.show()
```

**Figure 4.** Mean percentage change in MRI fat fraction and four-stair ascent time between T1 and T2 for 24 subjects. MRI showed a larger average change than four-stair ascent. Error bars represent standard deviations, indicating substantial variability in both measures.

## Conclusions:

Our analysis suggests that MRI fat fraction may be more sensitive to FSHD progression than four-stair ascent time. As shown in Figure 4, MRI fat fraction increased by an average of 4.88% (SD = 10.25%), compared to 0.28% (SD = 10.44%) for four-stair ascent across 24 subjects. Although MRI showed a larger average change, both measurements had substantial variability, making it difficult to determine which was more consistent.

Figures 3 and 5 also demonstrate that MRI fat fraction varies considerably across locations along the tibialis anterior muscle. This suggests that MRI could be a useful biomarker for detecting regional differences in muscle composition that functional assessments cannot directly measure. However, by comparison, it's possible that four-stair ascent could provide information about physical performance and mobility which may be more directly relevant to patients' daily lives. Additionally, MRI's cost and limited accessibility could prevent some patients from participating in trials, making it important to include functional assessments that tend to be more accessible, affordable, and directly relevant to the lives of patients.

Based on these preliminary findings, we would recommend prioritizing MRI fat fraction as an exploratory endpoint in early Phase 1/2 clinical trials, while including four-stair ascent as a complementary endpoint. Using both measurements could provide a more comprehensive understanding of FSHD progression and potential treatment benefits. However, before these measurements could be used to evaluate treatment effectiveness, further studies would be needed in order to confirm their reliability across larger patient populations and determine whether changes in MRI fat fraction correspond to relevant improvements in physical function.

## Limitations and Future Work:

One of our biggest limitations was the relatively small dataset of 34 subjects, with only 24 having complete interpolated measurements for both MRI fat fraction and four-stair ascent (Figure 4). Subjects also had limited visits, missing data, and inconsistent follow-up times. Additionally, linear and quadratic interpolation assume that there is smooth disease progression between visits, which is oftentimes not an accurate representation of what actually occurs.

Another limitation was that the spatial MRI analysis was limited to the tibialis anterior muscle of one subject. Figures 3 and 5 demonstrate substantial differences across muscle locations, meaning that individual slices may not represent the entire muscle. Errors in segmentation could also affect measurements, while functional assessments may vary due to factors that are unrelated to disease progression, such as time of day or fatigue level. Some subjects also showed decreases in MRI fat fraction or four-stair ascent time rather than the expected increases, suggesting individual variability, measurement uncertainty, or limitations in our interpolation methods.

In the future, studies should include more participants, additional visits and muscles, improved segmentation, and comparisons with other biomarkers and functional assessments. Further studies would also need to establish whether MRI changes correspond to meaningful functional improvements or treatment responses. Finally, MRI's cost and limited accessibility raise ethical concerns, while relying on imaging alone could overlook changes that matter to patients.
